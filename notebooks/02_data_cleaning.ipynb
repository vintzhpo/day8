```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/B_customer_churn.csv")
```

```python
df["monthly_spend"] = df["monthly_spend"].fillna(df["monthly_spend"].mean())
df["monthly_usage_gb"] = df["monthly_usage_gb"].fillna(df["monthly_usage_gb"].mean())
print(df.isna().sum())
```

```python
df["plan_type"] = df["plan_type"].str.strip().str.lower()
print(df["plan_type"].unique())
```

```python
print("Өмнөх давхардал:", df.duplicated().sum())
df = df.drop_duplicates()
print("Цэвэрлэсний дараах давхардал:", df.duplicated().sum())
```

```python
df.loc[df["tenure_months"] < 0, "tenure_months"] = np.nan
df.loc[df["monthly_spend"] > 150, "monthly_spend"] = np.nan
```

```python
df["tenure_months"] = df["tenure_months"].fillna(df["tenure_months"].mean())
df["monthly_spend"] = df["monthly_spend"].fillna(df["monthly_spend"].mean())
```

```python
df.to_csv("../data/B_customer_churn_cleaned.csv", index=False)
print("Дата амжилттай хадгалагдлаа. Нийт мөрний тоо:", len(df))
```
