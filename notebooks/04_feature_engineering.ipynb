```python
import pandas as pd 
import numpy as np 

df = pd.read_csv("../data/B_customer_churn_cleaned.csv")

df['tenure_bucket'] = pd.cut(df['tenure_months'], bins=[0, 6, 12, 24, 48, 72])

df['usage_per_spend'] = df['monthly_usage_gb'] / df['monthly_spend'].replace(0, np.nan)

df['complaint_bucket'] = pd.cut(df['complaints_count'], bins=[-1, 0, 2, 100], labels=['none', 'low', 'high'])

df["payment_recency_bucket"] = pd.cut(df["days_since_last_payment"], bins=[-1, 7, 30, 60])

df = pd.get_dummies(df, columns=['plan_type'], prefix='plan', drop_first=True)
df.to_csv('../data/features_df.csv', index=False)
print("feature-үүдийн амжилттай үүсгэгдлээ")
```
